**Program:**

```python
import numpy as np
import pandas as pd
import seaborn as sns
```


**Data Collection:**

```python
dt=pd.read_csv('/content/archive (2).zip')
```

**Data Preprocessing:**

```python
dt.drop_duplicates(inplace=True)
```

```python
dt.head()
```

```python
dt.label
```

**Label Encoding:**

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
dt.label=le.fit_transform(dt.label)
```

```python
dt.head()
```

**X,Y Splitup**

```python
x=dt.drop(columns=['N','P','ph' , 'S','Cu','Fe','Mn','label'])
```

```python
y=dt.label
```

**Train,Test,Split**

```python
from sklearn.model_selection import train_test_split
```

```python
xtr, xte, ytr, yte = train_test_split(x,y,test_size=0.15)
```

**Algorithm Decision**- Logistic Regression


When independent linear clusters are found

```python
from sklearn.linear_model import LogisticRegression
lr= LogisticRegression()
lr=lr.fit(xtr,ytr)
ypred=lr.predict(xte)
```

**Evaluation matrix**

```python
from sklearn.metrics import accuracy_score,roc_auc_score, confusion_matrix
```

```python
accuracy_score(yte,ypred)
```

```python
a=confusion_matrix(yte,ypred)
a
```

**KNN**- When a group of clusters are separated by a good distance


```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
k=KNeighborsClassifier(n_neighbors=5)
```

```python
k.fit(xtr,ytr)
ypred=k.predict(xte)
print(accuracy_score(yte,ypred))
print(confusion_matrix(yte,ypred))
```
